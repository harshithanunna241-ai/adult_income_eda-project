```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
import urllib.request
import os

# Create the raw data folder
os.makedirs("../data/raw", exist_ok=True)

# UCI Adult dataset URLs
train_url = "https://archive.ics.uci.edu/ml/machine-learning-databases/adult/adult.data"
test_url = "https://archive.ics.uci.edu/ml/machine-learning-databases/adult/adult.test"

# Download the files
urllib.request.urlretrieve(train_url, "../data/raw/adult.data")
urllib.request.urlretrieve(test_url, "../data/raw/adult.test")

print("Dataset downloaded successfully!")
```

```python
columns = [
    "age",
    "workclass",
    "fnlwgt",
    "education",
    "education_num",
    "marital_status",
    "occupation",
    "relationship",
    "race",
    "sex",
    "capital_gain",
    "capital_loss",
    "hours_per_week",
    "native_country",
    "income"
]

train_df = pd.read_csv(
    "../data/raw/adult.data",
    header=None,
    names=columns,
    skipinitialspace=True
)

print("Training data loaded successfully!")
print("Shape:", train_df.shape)

train_df.head()
```

```python
test_df = pd.read_csv(
    "../data/raw/adult.test",
    header=None,
    names=columns,
    skipinitialspace=True,
    skiprows=1
)

print("Test data loaded successfully!")
print("Shape:", test_df.shape)

test_df.head()
```

```python
train_df["source_split"] = "train"
test_df["source_split"] = "test"

df = pd.concat([train_df, test_df], ignore_index=True)

print("Both datasets combined successfully!")
print("Shape:", df.shape)
```

```python
df.head()
```

```python
df.info()
```

```python
# Check for the '?' missing-value marker
for column in df.columns:
    count = (df[column] == "?").sum()
    if count > 0:
        print(column, ":", count)
```

```python
# Replace '?' with NaN
df = df.replace("?", np.nan)

print(" '?' values replaced with NaN")
```

```python
missing_values = df.isnull().sum()

print(missing_values[missing_values > 0])
```

```python
# Replace missing categorical values with "Unknown"
categorical_columns = [
    "workclass",
    "occupation",
    "native_country"
]

for column in categorical_columns:
    df[column] = df[column].fillna("Unknown")

print("Missing categorical values replaced with 'Unknown'")
```

```python
df[categorical_columns].isnull().sum()
```

```python
df.head()
```

```python
df.info
```

```python
# Remove the trailing '.' from income values
df["income"] = df["income"].str.replace(".", "", regex=False)

print(df["income"].value_counts())
```

```python
# Create a numerical income flag
df["high_income"] = df["income"].map({
    "<=50K": 0,
    ">50K": 1
})

print(df[["income", "high_income"]].head())
```

```python
numeric_columns = [
    "age",
    "fnlwgt",
    "education_num",
    "capital_gain",
    "capital_loss",
    "hours_per_week"
]

for column in numeric_columns:
    df[column] = pd.to_numeric(df[column], errors="coerce")

print(df[numeric_columns].dtypes)
```

```python
print(df[numeric_columns].isnull().sum())
```

```python
print("Minimum age:", df["age"].min())
print("Maximum age:", df["age"].max())

print("Minimum hours:", df["hours_per_week"].min())
print("Maximum hours:", df["hours_per_week"].max())

print("Minimum education_num:", df["education_num"].min())
print("Maximum education_num:", df["education_num"].max())
```

```python
duplicate_count = df.duplicated().sum()

print("Number of duplicate rows:", duplicate_count)
```

```python
df = df.drop_duplicates()
```

```python
df["age_group"] = pd.cut(
    df["age"],
    bins=[0, 25, 35, 45, 55, 65, 100],
    labels=["18-25", "26-35", "36-45", "46-55", "56-65", "66+"]
)

print(df["age_group"].value_counts().sort_index())
```

```python
df["hours_group"] = pd.cut(
    df["hours_per_week"],
    bins=[0, 20, 40, 60, 100],
    labels=["0-20", "21-40", "41-60", "61+"]
)

print(df["hours_group"].value_counts().sort_index())
```

```python
df["net_capital"] = df["capital_gain"] - df["capital_loss"]

df[["capital_gain", "capital_loss", "net_capital"]].head()
```

```python
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```

```python
df.to_csv(
    "../data/raw/adult_income_raw.csv",
    index=False
)

print("Raw combined dataset saved successfully!")
```
